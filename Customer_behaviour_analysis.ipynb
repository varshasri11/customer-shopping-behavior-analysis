```python
import pandas as pd
```

```python
df = pd.read_csv('customer_shopping_behavior.csv')
```

```python
df.head()
```

```python
df.info()
```

```python
df.describe(include = 'all')
```

```python
#identifying null values
df.isnull().sum()
```

```python
#replacing null values with meaian
df['Review Rating'] = df.groupby('Category')['Review Rating'].transform(lambda x: x.fillna(x.median()))
```

```python
df.isnull().sum()
```

```python
#changing all the colm names into snakecase(lower case letters and spaces with _)
df.columns = df.columns.str.lower()
df.columns = df.columns.str.replace(' ','_')
```

```python
df.info()
```

```python
#chnaging purchase_amount_(usd) colm also into neat format
df = df.rename(columns = {'purchase_amount_(usd)':'purchase_amount'})
```

```python
df.info()
```

```python
#creating a new colm called "age_groups" to categorize age of customers into 4 diff categoeries
lables = ['Young Adults', 'Adults', 'Middle-aged','Senior']
df['age_group'] = pd.qcut(df['age'], q = 4, labels = lables)
```

```python
df[['age','age_group']].head(10)
```

```python
#finding unique values of freq_of_purchases colm
#unique() fun can be used of both numerical and text comls to get unique values of the colm
df['frequency_of_purchases'].unique()
```

```python
#finding number of unique values in that colm
df['frequency_of_purchases'].nunique()
```

```python
#creating a new colm and replacing those text values of "frequency_of_purchases" colm with numbers
#creating a map
freq_mapping = {'Weekly' : 7, 
                'Fortnightly': 14, 
                'Bi-Weekly': 14, 
                'Monthly': 30,
                'Quarterly':90, 
                'Annually': 365,
                'Every 3 Months':90 }
df['purchase_frequency_days'] = df['frequency_of_purchases'].map(freq_mapping)
```

```python
df[['frequency_of_purchases','purchase_frequency_days']].head(10)
```

```python
#inspecting both colms("discount_applied", "promo_code_used")
df['discount_applied'] == df['promo_code_used']
#the above code will returns T/F for all the rows
```

```python
df[['discount_applied','promo_code_used']].head(10)
```

```python
df[['discount_applied','promo_code_used']].dtypes
```

```python
(df['discount_applied'] == df['promo_code_used']).all()
```

```python
#dropping this colm because if promo code is used then discount will be aplied automatically. So, it doesnt make sense if we keep both colms 
df = df.drop('promo_code_used',axis = 1)
```

```python
import os
import pandas as pd
from sqlalchemy import create_engine, text

# 1. MySQL connection details
username = "root"
password = os.environ.get("DB_PASSWORD")  # never hardcode credentials
host = "localhost"
port = "3306"

# 2. Connect to MySQL (without a database)
engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}"
)

# 3. Create the database
with engine.connect() as conn:
    conn.execute(text("CREATE DATABASE IF NOT EXISTS mydatabase"))

print("Database created successfully!")

# 4. Connect to the new database
database = "mydatabase"

engine = create_engine(
    f"mysql+pymysql://{username}:{password}@{host}:{port}/{database}"
)

print("Connected to mydatabase!")

# 5. Write DataFrame to MySQL
table_name = "mytable"

df.to_sql(
    table_name,
    engine,
    if_exists="replace",
    index=False
)

print("DataFrame successfully written to MySQL!")

# 6. Read data back from MySQL
result = pd.read_sql(
    "SELECT * FROM mytable LIMIT 10;",
    engine
)

result
```
